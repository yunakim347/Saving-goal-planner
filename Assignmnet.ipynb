```python
pocketmoney=int(input("부모님으로 부터 받은 용돈: "))
change = pocketmoney - 3550
print(f"거스름돈: {change}원")
a= change//500
print(f"500원 짜리 거스름돈 : {a}개")
b= (change-a*500)//100
print(f"100원 짜리 거스름돈 : {b}개")
c= (change-(a*500+b*100))//50
print(f"50원 짜리 거스름돈 : {c}개")

```
